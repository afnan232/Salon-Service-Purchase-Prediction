# Salon Service Purchase Prediction

This project explores customer purchase prediction using demographic and visit-related features. It contains the original project workflow, Lab 2, and Lab 3 in one notebook.

**Target:** `Purchased` (0 = No, 1 = Yes).

> Note: The dataset is a small illustrative sample of 8 records, so model metrics are for demonstration and should not be treated as reliable real-world performance.

## Part 1 — Main Project: Initial Classification Workflow

```python
import pandas as pd

# STUDENTS: EDIT HERE
PROJECT_NAME = "Salon Service Purchase Prediction"
TARGET_COLUMN = "Purchased"

data = {
"Age": [22, 25, 30, 35, 28, 40, 21, 32],
"Visits": [2, 10, 5, 12, 3, 15, 1, 8],
"Previous_Purchases": [0, 3, 1, 5, 0, 6, 0, 3],
"Purchased": [0, 1, 0, 1, 0, 1, 0, 1]
}

# NO CHANGES NEEDED
df = pd.DataFrame(data)
print("Project:", PROJECT_NAME)
print("\nFirst Records:\n", df.head())
print("\nShape:", df.shape)
print("\nMissing Values:\n", df.isnull().sum())
print("\nTarget Distribution:\n", df[TARGET_COLUMN].value_counts())
```

```python
from sklearn.model_selection import train_test_split

# NO CHANGES NEEDED
X = df.drop(columns=[TARGET_COLUMN])
y = df[TARGET_COLUMN]
X_train, X_test, y_train, y_test = train_test_split(
X, y, test_size=0.25, random_state=42,
stratify=y
)
print("Training Records:", len(X_train))
print("Testing Records:", len(X_test))
print(" Data split successfully!")
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import make_pipeline

# NO CHANGES NEEDED
preprocess = make_pipeline(
SimpleImputer(strategy="median"),
StandardScaler()
)
X_train = preprocess.fit_transform(X_train)
X_test = preprocess.transform(X_test)
print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
print(" Data preprocessing completed!")
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier

# NO CHANGES NEEDED
model1 = LogisticRegression()
model2 = RandomForestClassifier(random_state=42)
model1.fit(X_train, y_train)
model2.fit(X_train, y_train)
print(" Logistic Regression trained!")
print(" Random Forest trained!")
```

```python
from sklearn.metrics import accuracy_score, f1_score, roc_auc_score, confusion_matrix

# NO CHANGES NEEDED
models = {
"Logistic Regression": model1,
"Random Forest": model2
}

for name, model in models.items():
    predictions = model.predict(X_test)
    probabilities = model.predict_proba(X_test)[:, 1]

    print("\nModel:", name)
    print("Accuracy:", round(accuracy_score(y_test, predictions), 2))
    print("F1 Score:", round(f1_score(y_test, predictions, zero_division=0), 2))
    print("ROC-AUC:", round(roc_auc_score(y_test, probabilities), 2))
    print("Confusion Matrix:\n", confusion_matrix(y_test, predictions))
    print("\n✅ Evaluation completed!")
```

## Part 2 — Lab 2: Feature Engineering and Pipeline

```python
import pandas as pd
PROJECT_NAME = "Salon Service Purchase Prediction"
TARGET_COLUMN = "Purchased"
data = {
"Age": [22, 25, 30, 35, 28, 40, 21, 32],
"Visits": [2, 10, 5, 12, 3, 15, 1, 8],
"City": ["Riyadh", "Jeddah", "Dammam", "Riyadh",
"Jeddah", "Dammam", "Riyadh", "Jeddah"],
"Purchased": [0, 1, 0, 1, 0, 1, 0, 1]
}
df = pd.DataFrame(data)
print("📌 Project:", PROJECT_NAME)
print("\n📋 First Records:\n", df.head())
print("\n📊 Dataset Shape:", df.shape)
print("\n🔍 Missing Values:\n", df.isnull().sum())
print("\n🎯 Target:", TARGET_COLUMN)
print("\n✅ Lab 2 — Code 1 completed!")
```

```python
# 🚀 Lab 2 — Code 2: Train / Test Split
from sklearn.model_selection import train_test_split
X = df.drop(columns=[TARGET_COLUMN])
y = df[TARGET_COLUMN]
X_train, X_test, y_train, y_test = train_test_split(
X, y,
test_size=0.25,
random_state=42,
stratify=y
)
print("📊 Total Records:", len(df))
print("📘 Training Records:", len(X_train))
print("📗 Testing Records:", len(X_test))
print("\n🎯 Training Target:\n", y_train.value_counts())
print("\n🎯 Testing Target:\n", y_test.value_counts())
print("\n✅ Lab 2 — Code 2 completed!")
```

```python
# 🚀 Lab 2 — Code 3: Feature Engineering
COLUMN_1 = "Visits"
COLUMN_2 = "Age"
NEW_FEATURE = "Visits_Per_Age"
X_train = X_train.copy()
X_test = X_test.copy()
for dataset in [X_train, X_test]:
    dataset[NEW_FEATURE] = (
        dataset[COLUMN_1] / (dataset[COLUMN_2] + 1)
    )
print("🛠️ New Feature:", NEW_FEATURE)
print("\n📋 Updated Data:\n", X_train.head())
print("\n📊 Total Features:", X_train.shape[1])
print("\n✅ Lab 2 — Code 3 completed!")
```

```python
# 🚀 Lab 2 — Code 4: Build Pipeline & Train
from sklearn.pipeline import make_pipeline
from sklearn.compose import make_column_transformer, make_column_selector
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.linear_model import LogisticRegression
prep = make_column_transformer(
(SimpleImputer(strategy="median"),
make_column_selector(dtype_include="number")),
(make_pipeline(SimpleImputer(strategy="most_frequent"),
OneHotEncoder(handle_unknown="ignore")),
make_column_selector(dtype_include="object"))
)
model = make_pipeline(prep, LogisticRegression(max_iter=1000))
model.fit(X_train, y_train)
print("🔧 Pipeline: Ready")
print("📊 Training Records:", len(X_train))
print("🧠 Model: Logistic Regression")
print("✅ Lab 2 — Code 4 completed!")
```

```python
# 🚀 Lab 2 — Code 5: Evaluate & Interpret
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix
predictions = model.predict(X_test)
print("🎯 Accuracy:", round(accuracy_score(y_test, predictions), 2))
print("\n📊 Classification Report:")
print(classification_report(y_test, predictions, zero_division=0))
print("\n🔍 Confusion Matrix:")
print(confusion_matrix(y_test, predictions))
print("\n✅ Lab 2 — Code 5 completed!")
```

## Part 3 — Lab 3: Model Comparison, Tuning, and Feature Importance

```python
import pandas as pd

PROJECT_NAME = "Salon Service Purchase Prediction"

df = pd.DataFrame({
"Age": [22, 25, 30, 35, 28, 40, 21, 32],
"Visits": [2, 10, 5, 12, 3, 15, 1, 8],
"City": ["Riyadh", "Jeddah", "Dammam", "Riyadh",
"Jeddah", "Dammam", "Riyadh", "Jeddah"],
"Purchased": [0, 1, 0, 1, 0, 1, 0, 1]
})

df["Visits_Per_Age"] = df["Visits"] / (df["Age"] + 1)

print("Project:", PROJECT_NAME)
print("Data Shape:", df.shape)
print(df.head())
print("✅ Lab 3 — Code 1 completed!")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import average_precision_score

X = pd.get_dummies(df.drop(columns=["Purchased"]))
y = df["Purchased"]

X_train, X_test, y_train, y_test = train_test_split(
X, y, test_size=0.25, random_state=42, stratify=y
)

baseline = RandomForestClassifier(random_state=42)
baseline.fit(X_train, y_train)

scores = baseline.predict_proba(X_test)[:, 1]
baseline_score = average_precision_score(y_test, scores)

print("📘 Training Records:", len(X_train))
print("📗 Testing Records:", len(X_test))
print("🌳 Model: Random Forest")
print("📊 Baseline PR-AUC:", round(baseline_score, 2))
print("✅ Lab 3 — Code 2 completed!")
```

```python

from sklearn.ensemble import GradientBoostingClassifier

boost = GradientBoostingClassifier(random_state=42)
boost.fit(X_train, y_train)

scores = boost.predict_proba(X_test)[:, 1]
boost_score = average_precision_score(y_test, scores)

print("🌳 Baseline PR-AUC:", round(baseline_score, 2))
print("🚀 Boosting PR-AUC:", round(boost_score, 2))
print("✅ Lab 3 — Code 3 completed!")
```

```python

from sklearn.model_selection import GridSearchCV

params = {
"n_estimators": [20, 50],
"learning_rate": [0.05, 0.1],
"max_depth": [1, 2]
}

search = GridSearchCV(
GradientBoostingClassifier(random_state=42),
params,
scoring="average_precision",
cv=2
)

search.fit(X_train, y_train)

best_model = search.best_estimator_

print("🔧 Best Parameters:", search.best_params_)
print("📊 Best CV PR-AUC:", round(search.best_score_, 2))
print("✅ Lab 3 — Code 4 completed!")
```

```python

from sklearn.metrics import f1_score, confusion_matrix

probabilities = best_model.predict_proba(X_test)[:, 1]

threshold = 0.5
predictions = (probabilities >= threshold).astype(int)

print("🎯 Threshold:", threshold)

print("📊 PR-AUC:",
round(average_precision_score(y_test, probabilities), 2))

print("📈 F1 Score:",
round(f1_score(y_test, predictions, zero_division=0), 2))

print("🔍 Confusion Matrix:\n",
confusion_matrix(y_test, predictions))

print("✅ Lab 3 — Code 5 completed!")
```

```python

import pandas as pd

importance = pd.Series(
best_model.feature_importances_,
index=X_train.columns
).sort_values(ascending=False)

print("📊 Feature Importance:")
print(importance.round(2))

print("🏆 Most Important Feature:", importance.index[0])

print("✅ Lab 3 — Code 6 completed!")
```
